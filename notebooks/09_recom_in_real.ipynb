# 추천 시스템 구축에서의 이슈

# 9.1. 신규 사용자와 아이템 (Cold Start Problem)

CF는 데이터가 없는 사용자나 아이템에 대해 좋은 추천을 만들어내기 어렵다. 

그래서 데이터가 없는 사용자나 아이템에 대해 어떤 식으로 추천할지 결정해야한다. 

과거, 보통은 Best Seller를 추천하는 방식을 사용. 데이터가 충분해지면 CF를 사용하는 방식.

그러면 아이템 추천도 데이터가 충분해야 CF 알고리즘의 선택을 받게 되는데.. 이 경우, 빈익빈 부익부 현상이 발생.

그래서 이 경우, 아이템은 데이터가 부족할 경우 추천 리스트에 포함시키는 방식을 사용하기도 함.

# 9.2. 확장성 (Scalability)

실제로는 수백억개의 아이템, 유저를 처리해야하는 경우가 있기에 "효율적인 계산"이 필수적.

대표적인 기술로는, 
- 병렬 처리: Spark 사용
- 배치 스케줄: 서비스 사용량이 낮은 새벽 시간대에 배치를 통한 모델 생성
- 알고리즘 변경: UB보다 IB를 사용한다던가
- 샘플링 모델 학습: 대표적인 사용자와 아이템만 학습하는 방법

# 9.3. 추천의 활용 (Presentation)

- 직접적 추천 리스트 제공
- 간접적 추천 리스트 제공 (고객이 보는 페이지 사이드에 배치하는 방식)
- 검색 결과 정렬 (고객 예상 선호도에 따라 검색 정렬을 변경)
- 맞춤형 광고 제공

# 9.5. 사용자의 간접 평가 데이터 확보 (Indirect Evaluation Data)

사용자들은 리뷰나 후기를 남기는 것에 대해 달가워 하지 않음...

직접적인 데이터를 모으기 어려울 경우, 갑접적인 데이터를 모으기 위한 노력이 필요.

가령, 쇼핑몰에서 어떤 페이지에 방문했고 체류 시간이 어떠한지 등.
