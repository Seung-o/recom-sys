# 추천 시스템 시작하기

이 노트북은 추천 시스템 프로젝트를 시작하기 위한 기본 설정과 샘플 코드를 포함합니다.


## 1. 라이브러리 임포트


```python
# 기본 라이브러리
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# 설정
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
%matplotlib inline

# 경고 무시
import warnings
warnings.filterwarnings('ignore')

print("✅ 라이브러리 임포트 완료")
```

## 2. 데이터 경로 설정


```python
# 프로젝트 경로 설정
PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / 'data'
MODEL_DIR = PROJECT_ROOT / 'models'

print(f"프로젝트 루트: {PROJECT_ROOT}")
print(f"데이터 폴더: {DATA_DIR}")
print(f"모델 폴더: {MODEL_DIR}")

# 폴더 생성 (존재하지 않는 경우)
DATA_DIR.mkdir(exist_ok=True)
MODEL_DIR.mkdir(exist_ok=True)
```

## 3. 샘플 데이터 생성

실습을 위한 간단한 샘플 데이터를 생성합니다.


```python
# 샘플 평점 데이터 생성
np.random.seed(42)

n_users = 100
n_items = 50
n_ratings = 1000

sample_data = pd.DataFrame({
    'user_id': np.random.randint(1, n_users + 1, n_ratings),
    'item_id': np.random.randint(1, n_items + 1, n_ratings),
    'rating': np.random.randint(1, 6, n_ratings),
    'timestamp': pd.date_range('2023-01-01', periods=n_ratings, freq='H')
})

# 중복 제거
sample_data = sample_data.drop_duplicates(subset=['user_id', 'item_id'])

print(f"생성된 데이터 크기: {sample_data.shape}")
sample_data.head()
```

## 4. 데이터 탐색


```python
# 기본 통계
print("=== 데이터 기본 정보 ===")
print(f"총 사용자 수: {sample_data['user_id'].nunique()}")
print(f"총 아이템 수: {sample_data['item_id'].nunique()}")
print(f"총 평점 수: {len(sample_data)}")
print(f"\n평점 분포:")
print(sample_data['rating'].value_counts().sort_index())
```

```python
# 평점 분포 시각화
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 평점 분포
sample_data['rating'].value_counts().sort_index().plot(kind='bar', ax=axes[0], color='skyblue')
axes[0].set_title('평점 분포', fontsize=14, fontweight='bold')
axes[0].set_xlabel('평점')
axes[0].set_ylabel('개수')
axes[0].grid(axis='y', alpha=0.3)

# 사용자별 평점 수
user_counts = sample_data['user_id'].value_counts()
user_counts.plot(kind='hist', bins=30, ax=axes[1], color='coral')
axes[1].set_title('사용자별 평점 수 분포', fontsize=14, fontweight='bold')
axes[1].set_xlabel('평점 수')
axes[1].set_ylabel('사용자 수')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. 간단한 추천 시스템 예제

인기도 기반 추천 (가장 기본적인 방법)


```python
# 아이템별 평균 평점과 평점 수 계산
item_stats = sample_data.groupby('item_id').agg({
    'rating': ['mean', 'count']
}).round(2)

item_stats.columns = ['avg_rating', 'num_ratings']
item_stats = item_stats.reset_index()

# 최소 평점 수 필터링 (예: 5개 이상)
min_ratings = 5
popular_items = item_stats[item_stats['num_ratings'] >= min_ratings]

# 평균 평점 순으로 정렬
popular_items = popular_items.sort_values('avg_rating', ascending=False)

print(f"=== 인기 아이템 TOP 10 (최소 {min_ratings}개 평점) ===")
print(popular_items.head(10))
```

## 6. 데이터 저장

생성한 샘플 데이터를 저장합니다.


```python
# 데이터 저장
output_path = DATA_DIR / 'sample_ratings.csv'
sample_data.to_csv(output_path, index=False)
print(f"✅ 데이터가 저장되었습니다: {output_path}")
```
